```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,StandardScaler
from sklearn.model_selection import RandomizedSearchCV,GridSearchCV,StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,recall_score,precision_score
```

```python
pd.set_option("display.max_rows",None)
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python

#Change the dtype of Total charge
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
df["MultipleLines"] = df["MultipleLines"].replace({"No phone service":"No"})
df.info()
```

```python
service_cols = ["OnlineSecurity", "OnlineBackup", "DeviceProtection",
                "TechSupport", "StreamingTV", "StreamingMovies"]
df["num_of_services"] = (df[service_cols]=="Yes").sum(axis=1)
df["has_protection"] = (df["OnlineSecurity"]=="Yes") | (df["TechSupport"]=="Yes")
```

```python
df.head()
```

```python


df["Churn_num"] = df["Churn"].map({"Yes":1,"No":0})
columns_to_drop = ['customerID','Churn','Churn_num']

df = df.dropna()
X =df.drop(columns=columns_to_drop)
y=df["Churn_num"]



```

```python
from sklearn.feature_selection import VarianceThreshold

numColumns = X.select_dtypes(include="number").columns.to_list()
catColumns = X.select_dtypes(include="object").columns.to_list()

preprocessor = ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),numColumns),
        ("cat",OneHotEncoder(),catColumns),
    ]
)
```

```python
pipeline = Pipeline(
    steps=[
       ("preprocess",preprocessor),
       ("clf",LogisticRegression(solver="saga",random_state=42,class_weight="balanced",max_iter=5000)) 
    ]
)
param_dict = {
    "clf__C":[0.001,0.01,1,10],
    "clf__l1_ratio": [0, 0.5, 1],
    
}
cv = StratifiedKFold(5, shuffle=True, random_state=42)

model = GridSearchCV(
    estimator=pipeline,
    param_grid=param_dict,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,stratify=y,random_state=42)
```

```python

model.fit(X_train,y_train)
proba = model.best_estimator_.predict_proba(X_test)[:,1]
threshold = 0.45
y_predict = (proba >= threshold).astype(int)

accuracy = accuracy_score(y_test,y_predict)
recall  = recall_score(y_test,y_predict)
precision = precision_score(y_test,y_predict)

print(f"The accuracy of the model is {accuracy}")
print(f"The recall of the model is {recall}")
print(f"The precision of the model  is {precision}")
```

```python
from sklearn.metrics import classification_report,confusion_matrix

class_report = classification_report(y_test,y_predict)
conf_matrix = confusion_matrix(y_test,y_predict)


plt.figure(figsize=(6, 5))
sns.heatmap(
    conf_matrix,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Stayed", "Churned"],
    yticklabels=["Stayed", "Churned"],
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion matrix")
plt.show()
```

```python
best = model.best_estimator_
names = best.named_steps["preprocess"].get_feature_names_out()
coefs = pd.Series(best.named_steps["clf"].coef_[0], index=names)

coefs = coefs[coefs != 0].sort_values()          # drop features L1 zeroed out
print(pd.concat([coefs.head(8), coefs.tail(8)]))

top = pd.concat([coefs.head(8), coefs.tail(8)])
top.index = [i.split("__")[1] for i in top.index]
top.plot.barh(color=["#2b6cb0" if v < 0 else "#c05621" for v in top], title="Coefficients (+ = more churn)")
```
